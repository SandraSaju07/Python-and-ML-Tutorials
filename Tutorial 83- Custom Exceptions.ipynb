```python
def check_priority(priority):
    if priority < 0 or priority > 5:
        raise ValueError(f"priority must be 1 to 5, got {priority}")
    return priority
```

```python
print(check_priority(4))
print(check_priority(9))
```

```python
import json

class TriageError(Exception):
    pass

def parse(answer):
    try:
        return json.loads(answer)
    except json.JSONDecodeError:
        raise TriageError(f"model did not return JSON {answer!r}")
```

```python
try:
    parse("I am not sure how to sort this one.")
except TriageError as error:
    print(f"caught: {error}")
```

```python
class TriageError(Exception):
    def __init__(self, ticket_id, answer):
        super().__init__(f"ticket {ticket_id}: could not sort {answer!r}")
        self.ticket_id = ticket_id
```

```python
try:
    raise TriageError(4, "I am not sure")
except TriageError as error:
    print(error)
    print(f"Needs a person {error.ticket_id}")
```

```python
class TriageError(Exception):
    def __init__(self, ticket_id, customer, answer):
        super().__init__(f"ticket {ticket_id} {customer} could not sort {answer!r}")
        self.ticket_id = ticket_id
        self.customer = customer
```

```python
try:
    raise TriageError(6, "Amanda", "Bad request")
except Exception as e:
    print(e)
    print(f"needs a person: {e.ticket_id} {e.customer}")
```
