```python
def add_item(item, basket=[]):
    basket.append(item)
    return basket
```

```python
print(add_item("apple"))
print(add_item("lemon"))
print(add_item("orange"))  
# default argument is created once only for the ones which wont pass the argument, shareed by the ones not passing it
```

```python
# fix to eliminate this behaviour
def add_item(item, basket=None):
    if basket is None:
        basket = []
    basket.append(item)
    return basket
```

```python
print(add_item("apple"))
print(add_item("orange"))
print(add_item("lemon"))
```
