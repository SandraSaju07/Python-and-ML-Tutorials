```python
import os, tempfile
```

```python
path = os.path.join(tempfile.mkdtemp(), "log.txt")
path
```

```python
with open(path, "w") as file:
    for i in range(1, 6):
        file.write(f"line: {i}\n")
```

```python
def read_large_files(file_path):
    with open(file_path, "r") as file:
        for line in file:
            yield line.rstrip("\n")
```

```python
for line in read_large_files(path):
    print(line)
    if line == "line: 3":
        break
```

```python
for line in read_large_files(path):
    print(line)
    if line == "line: 5":
        break
```

```python
def token_streamer(file_path):
    with open(file_path, "r") as file:
        for line in file:
            line = line.strip("\n")
            line = line.split()
            for word in line:
                yield word
```

```python
for line in token_streamer(path):
    print(line)
```
