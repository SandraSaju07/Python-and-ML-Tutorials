```python
tickets = [
    {"id": 1, "customer": "Asha", "text": "I was charged twice for one order"},
    {"id": 2, "customer": "Ben", "text": "My parcel has not arrived"},
    {"id": 3, "customer": "Chen", "text": "Can I get a refund for the blue mug?"}
]

for ticket in tickets:
    print(f"Ticket ID: {ticket["id"]} Customer Name: {ticket["customer"]}")
```

```python
billing = []
for ticket in tickets:
    if "charged" in ticket["text"].lower() or "refund" in ticket["text"].lower():
        billing.append(ticket)

billing
```

```python
billing[0]["customer"]
```

```python
billing[1]["customer"]
```
