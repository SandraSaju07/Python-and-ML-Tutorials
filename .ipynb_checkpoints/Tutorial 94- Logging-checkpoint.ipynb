```python
import logging

logging.basicConfig(level=logging.INFO)
logging.info("a normal event")
```

```python
# from most to least serious
logging.debug("smallest detail, for finding a bug")
logging.info("a normal event")
logging.warning("something to look at")
logging.error("something failed")
logging.critical("the program cannot go on")
```

```python
import sys

logging.basicConfig(level=logging.INFO, stream=sys.stdout,
                   format="%(name)s %(levelname)s %(message)s")
logging.debug("connecting to the model")
logging.info("loaded 5 tickets")
logging.warning("ticket 4 needs a person")
```

```python
logger = logging.getLogger("triage")
logger.info("ticket 1 sorted as billing")
logger.warning("ticket 4 needs a person")
```

```python
import logging

logging.basicConfig(
    filename="run.log", filemode="w", level=logging.INFO,
    format="%(asctime)s %(name)s %(levelname)s %(message)s",
    datefmt="%Y-%m-%d %H:%M:%S",  
)

logging.getLogger("triage").info("sorted 5 tickets")
logging.getLogger("triage").warning("ticket 4 needs a person")

# Read and print the file contents
#print(open("run.log", "r").read())  # wont write since jupyter notebook has different handlers
```
