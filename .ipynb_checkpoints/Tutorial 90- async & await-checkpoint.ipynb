```python
# without async
import time

def ask_model_slowly(text):
    time.sleep(0.5)
    return f"answer to {text}"

start = time.perf_counter()
for text in ["ticket:1", "ticket:2", "ticket:3"]:
    print(ask_model_slowly(text))
print(f"took {time.perf_counter() - start:.1f} seconds")    
```

```python
import asyncio

async def ask_model_slowly(text):
    await asyncio.sleep(0.5)
    return f"answer to {text}"

async def main():
    # awaiting one a time in the loop is still sequential, no faster yet, will take 1.5 sec still
    # still awaits one ticket before starting the next
    for text in ["ticket:1", "ticket:2", "ticket:3"]:
        print(await ask_model_slowly(text))

start = time.perf_counter()
#asyncio.run(main())
await main()  # in jupyter notebook use await main() instead of asyncio.run(main())
print(f"took {time.perf_counter() - start:.1f} seconds")
```

```python
async def main():
    # never awaited so wont work
    answer = ask_model_slowly("ticket: 1")
    print(answer)

await main()
```

```python
async def main():
    answer = await ask_model_slowly("ticket: 1")
    print(answer)

await main()
```

```python
async def ask_model_slowly(text):
    await asyncio.sleep(1)
    print(f"answer to {text}")

async def main():
    for text in ["ticket1", "ticket2", "ticket3"]:
        await ask_model_slowly(text)

start = time.perf_counter()
await main()
print(f"took {time.perf_counter() - start:.1f} seconds")
```
