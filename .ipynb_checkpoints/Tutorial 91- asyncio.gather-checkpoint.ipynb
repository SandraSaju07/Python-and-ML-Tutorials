```python
import asyncio
import time

async def ask_model_slowly(text):
    await asyncio.sleep(1)
    return f"answer to {text}"
```

```python
async def main():
    texts = ["ticket: 1", "ticket: 2", "ticket: 3"]
    start = time.perf_counter()
    answers = await asyncio.gather(*[ask_model_slowly(text) for text in texts])
    print(answers)
    print(f"took {time.perf_counter() - start:.1f} seconds")
```

```python
await main()
```

```python
# Limit calls with a Semaphore, without limit send every call at once
limit = asyncio.Semaphore(2)

async def ask_politely(text):
    async with limit:  # wait for one of the places to be free, runs the block & gives back the place
        return await ask_model_slowly(text)

# 5 calls, 2 at atime, each takes 1 sec so total time => 2 calls + 2 calls + 1 call = 1 sec + 1 sec + 1 sec = 3 sec
async def main():
    start = time.perf_counter()
    answers = await asyncio.gather(*[ask_politely(f"ticket {i}") for i in range(1,6)])
    print(answers)
    print(f"took {time.perf_counter() - start:.1f}")
```

```python
await main()
```

```python
limit = asyncio.Semaphore(1)

async def ask_politely(text):
    async with limit:
        return await ask_model_slowly(text)

# 10 calls, one a time, each call 1 sec, so will take 10 sec
async def main():
    start = time.perf_counter()
    answers = await asyncio.gather(*[ask_politely(f"ticket: {i}") for i in range(1,11)])
    print(answers)
    print(f"took: {time.perf_counter() - start:.1f} sec")

await main()
```

```python
limit = asyncio.Semaphore(5)

async def ask_politely(text):
    async with limit:
        return await ask_model_slowly(text)

# 10 calls, 5 at a time, 1 sec each => 10 / 5 * 1 = 2 
async def main():
    start = time.perf_counter()
    answers = await asyncio.gather(*[ask_politely(f"ticket: {i}") for i in range(1,11)])
    print(answers)
    print(f"took: {time.perf_counter() - start:.1f}")

await main()
```
