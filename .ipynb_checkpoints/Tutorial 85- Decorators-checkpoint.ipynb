```python
def main_welcome(func):
    def sub_welcome_method():
        print("Welcome to advanced python course")
        func()
        print("Please learn these concepts clearly")
    return sub_welcome_method

def course_introduction():
    print("This is advanced Python course")
```

```python
greet = main_welcome(course_introduction)
greet()
```

```python
@main_welcome
def course_introduction():
    print("This is advanced Python course")
```

```python
def main_welcome(func):
    def sub_welcom_method():
        print("Welcome to advanced python course")
        func()
        print("Please learn properly")
    return sub_welcom_method()

@main_welcome
def course_introduction():
    print("this is advanced python course")
```

```python
def announce(func):
    def wrapper(*args, **kwargs):
        print("calling...", func.__name__)
        result = func(*args, **kwargs)
        print("done!")
        return result
    return wrapper

@announce
def add(a, b):
    return a + b
```

```python
add(2, 4)
```

```python
def repeat(times):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(times):
                func(*args, **kwargs)
        return wrapper
    return decorator
```

```python
@repeat(5)
def ping():
    print("ping")
```

```python
ping()
```

```python
def retry(attempts):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for attempt in range(1, attempts + 1):
                try:
                    return func(*args, **kwargs)
                except ValueError as error:
                    print(f"attempt {attempt} failed: {error}")
            raise RuntimeError(f"gave up after {attempts} attempts")
        return wrapper
    return decorator
```

```python
calls = {"n": 0}

@retry(attempts=3)
def flaky_call():
    calls["n"] += 1
    if calls["n"] < 3:
        raise ValueError("model times out")
    return "model replied"

print(flaky_call())
```
