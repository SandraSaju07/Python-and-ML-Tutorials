```python
from dataclasses import dataclass
```

```python
@dataclass
class Person:
    name: str
    age: int
    city: str

person = Person("Annu", 23, "NYC")
print(person)
```

```python
person = Person("Jian", 34, 23)
person
```

```python
class Person(BaseModel):
    name: str
    age: int
    city: str

person = Person(name="Kia", age=23, city="London")  # fields are passed by name here unlike dataclasses
person
```

```python
person = Person(name="Quinn", age=45, city=78)
person
```

```python
class Triage(BaseModel):
    category: str
    priority: int

t = Triage(category="billing", priority=4)
t
```

```python
t.priority + 1
```

```python
result = Triage(category="shipping", priority="5")
result
```

```python
type(result.priority)  # converted by pydantic since its clearly a number
```

```python
result = Triage(category="refund", priority="four")  # rejects value which cant convert
```

```python
from pydantic import Field

class Item(BaseModel):
    name: str = Field(min_length=1, max_length=10)
    price: float = Field(gt=0, le=1000)
    quantity: float = Field(ge=0)
```

```python
item = Item(name="Book", price=23, quantity=2)
item
```

```python
item = Item(name="Glass", price=-3, quantity=4)
item
```

```python
from typing import Literal
```

```python
class Triage(BaseModel):
    category: Literal["billing", "shipping", "other"]  # allows only the listed values
    priority: int = Field(gt=0, le=5)
```

```python
t =Triage(category="billing", priority=5)
t
```

```python
t = Triage(category="other", priority=6)
t
```

```python
result = Triage(category="billing", priority=4)
result.model_dump()
```

```python
class Triage(BaseModel):
    category: Literal["billing", "shipping", "other"]
    priority: int = Field(gt=1, le=5)
    reply: str | None = None
```

```python
result = Triage(category="billing", priority=3)
result
```

```python
result = Triage(category="other", priority=4.0)
result
```

```python
result = Triage(category="billing", priority=4.5)
result
```

```python
from pydantic import ValidationError
```

```python
try:
    result = Triage(category="billing", priority=4.5)
except ValidationError as e:
    print(e)
```
