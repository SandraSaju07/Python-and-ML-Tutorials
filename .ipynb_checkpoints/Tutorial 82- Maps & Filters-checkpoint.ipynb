```python
lst = [1,2,3,4,5,6,7,8,9,0]
list(map(lambda x: x%2 == 0, lst))
```

```python
list(map(lambda x: x*x, lst))
```

```python
list(filter(lambda x: x%2, lst))
```

```python
list(filter(lambda x: x*x, lst))
```

```python
tickets = ["Refund please ", "Parcel not delivered"]
list(map(str.strip, tickets))
```

```python
list(map(len, tickets))
```

```python
prompts, replies = [34, 20, 15], [12, 14, 8]
totals = list(map(lambda x,y: x+y, prompts, replies))
totals
```

```python
scores = [0.2, 0.55, 0.9, 0.4, 0.94]
list(filter(lambda x: x > 0.9, scores))
```

```python
tickets = ["Return the product", "Faulty item delivered", "Refund not recieved"]
list(map(str.upper, tickets))
```

```python
[x for x in scores if x > 0.9]
```
